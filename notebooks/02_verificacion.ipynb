```python
import glob, hashlib, re
from pathlib import Path
import pandas as pd

ruta_md5 = glob.glob("*.md5")[0]
resultados = []
with open(ruta_md5, encoding="utf-8-sig") as f:
    for linea in f:
        m = re.match(r"^([0-9A-Fa-f]{32})\s+\*?(.+?)\s*$", linea)
        if not m:
            continue
        esperado, ruta = m.group(1).upper(), m.group(2)
        nombre = ruta.replace("\\", "/").split("/")[-1]
        archivo = Path(nombre)
        if not archivo.exists():
            resultados.append((nombre, "NO SUBIDO", ""))
            continue
        h = hashlib.md5()
        with open(archivo, "rb") as g:
            for trozo in iter(lambda: g.read(1024 * 1024), b""):
                h.update(trozo)
        real = h.hexdigest().upper()
        resultados.append((nombre, "OK" if real == esperado else "FALLO", real))

pd.DataFrame(resultados, columns=["archivo", "resultado", "md5_calculado"])
```
